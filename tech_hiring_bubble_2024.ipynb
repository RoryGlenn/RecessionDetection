```python
# Reference Data: https://fredblog.stlouisfed.org/2023/03/was-there-a-tech-hiring-bubble/?utm_source=series_page&utm_medium=related_content&utm_term=related_resources&utm_campaign=fredblog

import os

import pandas as pd
import plotly.express as px
from fredapi import Fred

fred = Fred(api_key=os.environ["FRED_API_KEY"])
```

```python
total_job_postings_indeed = fred.get_series("IHLIDXUS")
total_job_postings_indeed = pd.DataFrame(
    total_job_postings_indeed, columns=["Job Postings on Indeed in the United States"]
)
```

```python
swde_jobs_indeed = fred.get_series("IHLIDXUSTPSOFTDEVE")
swde_jobs_indeed = pd.DataFrame(
    swde_jobs_indeed,
    columns=["Software Developer Job Postings on Indeed in the United States"],
)
```

```python
it_jobs_indeed = fred.get_series("IHLIDXUSTPITOPHE")
it_jobs_indeed = pd.DataFrame(
    it_jobs_indeed, columns=["IT Job Postings on Indeed in the United States"]
)
```

```python
df = pd.merge(
    total_job_postings_indeed,
    swde_jobs_indeed,
    how="inner",
    left_index=True,
    right_index=True,
)
df = pd.merge(df, it_jobs_indeed, how="inner", left_index=True, right_index=True)
df
```

```python
fig = px.line(
    df,
    x=df.index,
    y=df.columns,
    title="Job Postings on Indeed in the United States",
    template="plotly_dark",
    width=1000,
    height=400,
    labels={"value": "Job Postings", "variable": "Job Type", "index": "Date"},
    color_discrete_map={"Job Postings on Indeed in the United States": "white"},
)

fig.show()
```
